```python
import os, subprocess, sys, urllib.request, pathlib

print("python", sys.version.split()[0])
import torch
print("torch", torch.__version__, "| built cuda:", torch.version.cuda, "| cuda available:", torch.cuda.is_available())
print("device count:", torch.cuda.device_count())
print("/proc/driver/nvidia exists:", pathlib.Path("/proc/driver/nvidia").exists())
print("nvidia devices:", sorted(p.name for p in pathlib.Path("/dev").glob("nvidia*")) or "none")
print("image:", os.environ.get("KAGGLE_DOCKER_IMAGE", "?"))

for url in ("https://huggingface.co", "https://raw.githubusercontent.com/declare-lab/MELD/master/data/MELD/dev_sent_emo.csv"):
    try:
        request = urllib.request.Request(url, method="HEAD")
        with urllib.request.urlopen(request, timeout=20) as response:
            print("REACHABLE", response.status, url)
    except Exception as error:
        print("BLOCKED  ", type(error).__name__, error, url)
```
